````python
client = OpenAI(
    api_key=openai_api_key,
    base_url=openai_api_base,
)

models = client.models.list()
model = models.data[0].id

# Round 1
messages = [{"role": "user", "content": "9.11 and 9.8, which is greater?"}]
response = client.chat.completions.create(model=model, messages=messages)

reasoning = response.choices[0].message.reasoning
content = response.choices[0].message.content

print("reasoning for Round 1:", reasoning)
print("content for Round 1:", content)

# Round 2
messages.append({"role": "assistant", "content": content})
messages.append(
    {
        "role": "user",
        "content": "How many Rs are there in the word 'strawberry'?",
    }
)
response = client.chat.completions.create(model=model, messages=messages)

reasoning = response.choices[0].message.reasoning
content = response.choices[0].message.content

print("reasoning for Round 2:", reasoning)
print("content for Round 2:", content)
````

````python
from pydantic import BaseModel

class Info(BaseModel):
    name: str
    age: int

model = client.models.list().data[0].id
completion = client.beta.chat.completions.parse(
    model=model,
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What's my name and age?"},
    ],
    response_format=Info,
    max_completion_tokens=512,
)

message = completion.choices[0].message
print(message)
print("Name:", message.parsed.name)
print("Age:", message.parsed.age)
````

````python
model = client.models.list().data[0].id

completion = client.chat.completions.create(
    model=model,
    messages=[
        {"role": "system", "content": "Answer in json with fields {'name': your_name, 'age': 123}"},
        {"role": "user", "content": "Classify this sentiment: vLLM is wonderful!"}
    ],
    # extra_body={"structured_outputs": {"json": Info.model_json_schema()}},
    extra_body={"structured_outputs": {"regex": "name.*\sage.*"}},
    max_completion_tokens=600,
)
print(completion.choices[0].message.content)
````

````python
simplified_sql_grammar = """
    root ::= select_statement

    select_statement ::= "SELECT " column " from " table " where " condition

    column ::= "col_1 " | "col_2 "

    table ::= "table_1 " | "table_2 "

    condition ::= column "= " number

    number ::= "1 " | "2 "
"""

completion = client.chat.completions.create(
    model=model,
    messages=[
        {
            "role": "user",
            "content": "Generate an SQL query to show the 'username' and 'email' from the 'users' table.",
        }
    ],
    extra_body={"structured_outputs": {"grammar": simplified_sql_grammar}},
)
print(completion.choices[0].message.content)
````

````python
from pydantic import conlist, Field
from typing import Annotated

class Person(BaseModel):
    name: str
    age: Annotated[int, Field(ge=5, le=10)]

class Course(BaseModel):
    name: str = Field(description="Name of the course")
    persons: conlist(Person, min_length=3, max_length=5)
    description: str | None = None
````
