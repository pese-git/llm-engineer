```python
import json
from typing import Any
```

```python
def parse_message(msg: dict[str, Any]) -> dict[str, Any]:
    text_parts: list[str] = []
    for t in msg.get("text", []):
        if isinstance(t, str):
            text_parts.append(t)
        elif isinstance(t, dict):
            text_parts.append(t.get("text", ""))
    text = "".join(text_parts)

    reactions = {}
    for r in msg.get("reactions", []):
        if r["type"] == "emoji":
            name = r["emoji"]
        elif r["type"] == "custom_emoji":
            name = r["document_id"]
        elif r["type"] == "paid":
            name = "paid"
        else:
            name = r["type"]
        reactions[name] = r["count"]

    return {
        "from": msg.get("from"),
        "date": msg.get("date"),
        "text": text,
        "reactions": reactions
    }
```

```python
with open("aith_channel.json", "r", encoding="utf-8") as f:
    data = json.load(f)

parsed = []
for m in data["messages"]:
    parse = parse_message(m)
    if len(parse["text"].strip()) == 0:
        continue
    parsed.append(parse)
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("sergeyzh/rubert-mini-frida")
```

```python
messages = [p["text"] for p in parsed]
embeddings = model.encode(messages, show_progress_bar=True, batch_size=4)
```

```python
query_vector = model.encode(["Курс LLM-engeneer"])
```

```python
from qdrant_client import QdrantClient
from qdrant_client.models import VectorParams, PointStruct, Distance

# for local deploy with docker
# client = QdrantClient(host="localhost", port=6333)

client = QdrantClient(":memory:")
COLLECTION_NAME = "demo"


if not client.collection_exists(collection_name=COLLECTION_NAME):
    client.create_collection(
        collection_name=COLLECTION_NAME,
        vectors_config=VectorParams(size=embeddings.shape[1], distance=Distance.COSINE),
    )
```

```python
points = [
    PointStruct(id=i, vector=embedding, payload=metadata)
    for i, (embedding, metadata) in enumerate(zip(embeddings, parsed))
]
client.upsert(collection_name=COLLECTION_NAME, points=points)
```

```python
#%%timeit
hits = client.query_points(
    collection_name=COLLECTION_NAME,
    query=query_vector[0],
    limit=5,
)
```

```python
for hit in hits.points:
    print(hit.score, hit.payload)
```

# PGVector

```python
from sqlalchemy import (
    Index,
    JSON,
    String,
    select,
)
from sqlalchemy.orm import DeclarativeBase, Mapped, mapped_column, Session
from pgvector.sqlalchemy import Vector

class Base(DeclarativeBase):
    pass

class Document(Base):
    __tablename__ = "documents"

    id: Mapped[int] = mapped_column(primary_key=True, autoincrement=True)
    title: Mapped[str] = mapped_column(String, nullable=False)
    metadata_: Mapped[dict] = mapped_column(JSON, nullable=True)
    embedding: Mapped[list[float]] = mapped_column(Vector(d), nullable=False)

    __table_args__ = (
        Index(
            "documents_embedding_hnsw",
            "embedding",
            postgresql_using="hnsw",
            postgresql_ops={"embedding": "vector_l2_ops"},
            postgresql_with={"m": 16, "ef_construction": 200},
        ),
    )
```

```python
from sqlalchemy import create_engine, text

engine = create_engine("postgresql+psycopg://postgres:password@localhost/example_db")
with Session(engine) as session:
    with session.begin():
        session.execute(text("CREATE EXTENSION IF NOT EXISTS vector;"))

Base.metadata.create_all(engine)
```

```python
with Session(engine) as session:
    docs = [
        Document(
            title=p["text"],
            metadata=p,
            embedding=emb.tolist(),
        )
        for p, emb in zip(parsed, embeddings)
    ]
    session.add_all(docs)
    session.commit()
```

```python
with Session(engine) as session:
    stmt = (
        select(Document, Document.embedding.l2_distance(query_vector).label("dist"))
        .order_by("dist")
        .limit(5)
    )
    results = session.execute(stmt).all()
    for doc, dist in results:
        print(f"id={doc.id}, title={doc.title}, dist={dist:.6f}")
```

```python
from langchain.vectorstores import FAISS
from langchain.embeddings import HuggingFaceEmbeddings

model = HuggingFaceEmbeddings(
    model_name="sergeyzh/rubert-mini-frida",
#    model_kwargs={"device": "cpu"}
)
```
